```python
from metasyn.distribution.uniform import DurationUniformDistribution, DateTimeUniformDistribution
from metasyn.distribution import ColumnReference
from metasyn import MetaFrameBuilder
from metasyn.metaframe import DependencyGraph
```

```python
DurationUniformDistribution("P1D", "P2D").draw()
```

```python
builder = MetaFrameBuilder()
builder.add_column("start_time", var_type="datetime")
builder.add_column("end_time", var_type="datetime")
builder.n_rows = 20
builder["start_time"].distribution = DateTimeUniformDistribution("2015-07-23", "2015-05-08", precision="seconds")
builder["end_time"].distribution = ColumnReference("start_time") + DurationUniformDistribution("P1D", "P2D", precision="seconds")
mf = builder.fit()
```

```python
df = mf.synthesize(progress_bar=False)
```

```python
builder = MetaFrameBuilder()
builder.add_dataframe(df)
builder.add_column("duration", hidden=True)
builder["duration"].series = ColumnReference("end_time") - ColumnReference("start_time")
builder["end_time"].distribution = ColumnReference("start_time") + ColumnReference("duration")
builder.fit().synthesize()
```
