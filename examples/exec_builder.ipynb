```python

from metasyn.builder import MetaFrameBuilder, VarBuilder
from metasyn.privacy import BasicPrivacy
from metasyn.distribution import NADistribution, DiscreteUniformDistribution, StringConstantDistribution, ContinuousUniformDistribution, RegexDistribution, DateUniformDistribution
from metasyn.distribution.base import ColumnReference, EqualsCondition, IfThenElse, GreaterThanCondition, LessThanCondition, NotEqualsCondition
from metasyn import demo_data
from metasyncontrib.disclosure.privacy import DisclosurePrivacy
from pathlib import Path
import polars as pl
```

```python
df = demo_data("titanic")
df = df.with_columns(
    (pl.col("Age") > 18).alias("Grownup")
)
builder = MetaFrameBuilder()
builder.add_dataframe(df, None)
builder["PassengerId"].distribution = {"unique": True}
# builder["PassengerId"].distribution = ColumnReference("base")
builder["Fare"].distribution = {"name": "uniform"}
builder["Fare"].distribution = ContinuousUniformDistribution(lower=0.1, upper=0.2)
builder["Fare"].distribution = ContinuousUniformDistribution
builder["Fare"].distribution = None

builder["Fare"].privacy = BasicPrivacy()
builder.privacy = BasicPrivacy()
# builder.preview()
# builder["Age"].distribution = IfThenElse(ColumnReference("Sex") == "male", None, DiscreteUniformDistribution(20, 40))
builder["Age"].prop_missing = 0.1
builder["Fare"].distribution = ColumnReference("Age")
# builder["Sex"].distribution = ColumnReference("Name") + RegexDistribution(" (M|F)")
builder["Grownup"].distribution = GreaterThanCondition(ColumnReference("Age"), 18)
```
