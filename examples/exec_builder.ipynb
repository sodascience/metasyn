```python
from metasyn.builder import MetaFrameBuilder, VarBuilder, ColumnReference
from metasyn.privacy import BasicPrivacy
from metasyn.distribution import ContinuousUniformDistribution
from metasyn import demo_data
from metasyncontrib.disclosure.privacy import DisclosurePrivacy
```

```python
df = demo_data("titanic")
builder = MetaFrameBuilder()
builder.add_dataframe(df, None)
builder["PassengerId"].distribution = {"unique": True}
# builder["PassengerId"].distribution = ColumnReference("base")
builder["Fare"].distribution = {"name": "uniform"}
builder["Fare"].distribution = ContinuousUniformDistribution(lower=0.1, upper=0.2)
builder["Fare"].distribution = ContinuousUniformDistribution
builder["Fare"].distribution = None

builder["Fare"].privacy = BasicPrivacy()
builder["Fare"].distribution = {"prop_missing": 0.3}
builder.privacy = BasicPrivacy()
builder.preview()
```
